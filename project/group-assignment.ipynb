# DATASCI 350: group project random assignment

This notebook assigns students to groups of 4-5 for the final project.  
**Groups 01-07** were formed by students. **Groups 08-09** were randomly assigned (seed `350`).  
**Update (2026-03-25):** Fei Yi moved from Group 08 to Group 01 by request.

> **Note:** some emails were confirmed by students. All others follow
> the `firstname.lastname@emory.edu` pattern and **should be verified** against Canvas,
> as some students use middle names or different first names in their Emory email.

```python
import pandas as pd
from IPython.display import display, Markdown

# --- Known emails (confirmed by students) ---
known_emails = {
    "Minna Lai": "minna.lai@emory.edu",
    "Tianyi Zhou": "tianyi.grace.zhou@emory.edu",
    "Sarah Shao": "kexin.shao@emory.edu",
    "Serena Gu": "serena.gu@emory.edu",
    "Stan Xu": "stan.xu@emory.edu",
    "Yunshi Zhou": "yunshi.zhou@emory.edu",
    "Oliver Xie": "lxie46@emory.edu",
    "James Teng": "yteng27@emory.edu",
    "Bonnie Bao": "bonnie.bao@emory.edu",
}

def get_email(name):
    """Return known email or generate firstname.lastname@emory.edu."""
    if name in known_emails:
        return known_emails[name]
    return ".".join(name.lower().split()) + "@emory.edu"

# --- Pre-formed groups ---
group_01 = ["Fei Yi", "Minna Lai", "Sarah Shao", "Serena Gu", "Tianyi Zhou"]
group_02 = ["Christina Dai", "Eric Wang", "Joe Xia", "Kenneth He", "Tina Piltner"]
group_03 = ["Ayushi Chowksey", "Chau Anh Nguyen", "Hank Standaert", "Jacob Hillman", "Ryana Rajesh"]
group_04 = ["Bonnie Bao", "James Teng", "Oliver Xie", "Stan Xu", "Yunshi Zhou"]
group_05 = ["Eric Chen", "Steven Hu", "Yan Tian", "Yang Lyu"]
group_06 = ["Connor Lee", "Eshaan Dani", "Ihan Do", "Jonathan Luo", "Nikhil Makker"]
group_07 = ["Anny Huang", "Chrissy Wang", "Rola Ke", "Sissi Li", "Vicki Lin"]

# --- Groups originally assigned randomly (seed 350), now frozen ---
# Fei Yi moved from Group 08 to Group 01 by request (2026-03-25)
group_08 = ["Arush Singhania", "Saahil Mardhekar", "Theo Burns", "Vicente Martinez"]
group_09 = ["Hanna Florence", "Lucas Huang", "Selena Lin", "Zhenyu Zou"]

all_groups = [group_01, group_02, group_03, group_04, group_05,
              group_06, group_07, group_08, group_09]

n_pre = 7  # groups 01-07 were self-formed; 08-09 were random (now frozen)

print(f"Total students: {sum(len(g) for g in all_groups)}")
print(f"Self-formed: Groups 01-07")
print(f"Originally random (now frozen): Groups 08-09")
print(f"Total groups: {len(all_groups)}")
```

```python
# Display all groups
for i, group in enumerate(all_groups, 1):
    label = "(self-formed)" if i <= n_pre else "(random, frozen)"
    display(Markdown(f"### Group {i:02d} {label}"))
    df = pd.DataFrame(
        [{"Name": s, "Email": get_email(s)} for s in sorted(group)]
    )
    df.index += 1
    display(df)

# Summary table
display(Markdown("---\n### Summary"))
rows = []
for i, group in enumerate(all_groups, 1):
    formed = "self-formed" if i <= n_pre else "random"
    for s in group:
        rows.append({"Group": f"{i:02d}", "Name": s, "Email": get_email(s), "Type": formed})
summary = pd.DataFrame(rows).sort_values("Name").reset_index(drop=True)
summary.index += 1
display(summary)
```
